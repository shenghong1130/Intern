# 举左手后写入机器人身份测试

执行顺序：站立 → 左手举起 → 写入身份注册请求并等待 Start 写回任务 → 站立。执行最后一个单元后，请把机器人移到对应 Start 的读卡区。

## 1. 初始化依赖

```python
import time

from robotall import act, register_robot
```

## 2. 设置身份注册参数

```python
# 按本局比赛实际信息修改。
TEAM = 'red'
SECRET = '1234'
ROBOT_ID = 'red-1'

# 0 表示不限时间：写入请求后可从容地把机器人移动到 Start 读卡器。
SCAN_TIMEOUT_S = 0
```

## 3. 执行：举左手、写入身份、站立收尾

```python
act('stand')

try:
    print('1/3 左手举起')
    act('lift_left_hand', stand=False)
    time.sleep(0.5)

    print('2/3 写入机器人身份注册请求')
    print('现在将机器人移动到 Start 读卡区，等待服务器分配任务……')
    result = register_robot(
        team=TEAM,
        secret=SECRET,
        robot_id=ROBOT_ID,
        clear_first=True,
        read_response=True,
        wait_response=True,
        scan_timeout_s=SCAN_TIMEOUT_S,
        verbose_wait=True,
    )
    print(result)
    if not result['ok']:
        raise RuntimeError('没有收到有效的 Start 身份注册响应')
    print('3/3 身份注册流程完成')
finally:
    act('stand')
    print('已执行结束动作：stand')
```
