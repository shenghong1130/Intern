# 左手举起后换花测试

执行顺序：站立 → 左手举起 → 写入换花请求并等待 Worker 响应 → 站立收尾。执行换花单元后，再把机器人移动到对应 Worker 的读卡区。

## 1. 初始化依赖

```python
import time

from robotall import act, send_request
```

## 2. 设置本次换花参数

```python
# 按本局比赛实际信息修改这几项。
TEAM = 'red'
SECRET = '1234'
ROBOT_ID = 'red-1'
WORKER_ID = 12
FROM_FLOWER = 'main'       # Worker 当前显示的花色
TO_FLOWER = 'bailianhua'   # 注册时服务端分配给本队的目标花

# 0 表示不限时间：请求写入后，请把机器人移到 Worker 读卡器。
SCAN_TIMEOUT_S = 0
```

## 3. 执行：举左手、请求换花、站立收尾

```python
act('stand')

try:
    print('1/3 左手举起')
    act('lift_left_hand', stand=False)
    time.sleep(0.5)

    print(f'2/3 已写入换花请求：Worker {WORKER_ID}, {FROM_FLOWER} -> {TO_FLOWER}')
    print('现在将机器人移动到对应 Worker 的读卡区，等待判定结果……')
    result = send_request(
        team=TEAM,
        secret=SECRET,
        robot_id=ROBOT_ID,
        worker_id=WORKER_ID,
        from_flower=FROM_FLOWER,
        to_flower=TO_FLOWER,
        clear_first=True,
        read_response=True,
        wait_response=True,
        scan_timeout_s=SCAN_TIMEOUT_S,
        verbose_wait=True,
    )
    print(result)
    if not result['ok']:
        raise RuntimeError('没有收到有效的 Worker 换花响应')
    print('3/3 换花流程完成')
finally:
    act('stand')
    print('已执行结束动作：stand')
```
